```python
df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')
```

```python
df['Timestamp'] = pd.to_datetime(
    df['Date'].dt.strftime('%Y-%m-%d') + ' ' + df['Time'])
```

```python
df.head(2)
```

```python
df.columns
```

```python
df['Cancelled Rides by Customer']=df['Cancelled Rides by Customer'].fillna(0)
df['Cancelled Rides by Driver']=df['Cancelled Rides by Driver'].fillna(0)
df['Incomplete Rides']=df['Incomplete Rides'].fillna(0)
```

```python
d={'Auto':506.483049,'Bike':509.1145076,'eBike':503.4585559,'Go Mini':507.3814222,'Go Sedan':512.026865,
'Premier Sedan':509.5676324,'Uber XL':505.3021919}
```

```python
df['Estimated_Values'] = df['Vehicle Type'].map(d)
```

```python
df['Estimated_Values'] = df['Booking Value'].fillna(df['Vehicle Type'].map(d))
```

```python
failed_rides = df[df['Booking Status'] != 'Completed']
```
